## **Wprowadzenie do Regresji Liniowej**

### **Czym jest regresja liniowa?**

Regresja liniowa jest jednym z podstawowych modeli statystycznych, szeroko stosowanym do przewidywania wartości ciągłych. Jej głównym celem jest znalezienie prostej linii, która jak najdokładniej opisuje zależność pomiędzy zmienną niezależną $ x $ a zmienną zależną $ y $.

### **Równanie prostej regresji liniowej**

Regresję liniową można opisać równaniem prostej:

$$
y = \beta_0 + \beta_1 x + \epsilon
$$

gdzie:

- $ y $ jest zmienną zależną,
- $ x $ jest zmienną niezależną,
- $ \beta_0 $ jest wyrazem wolnym (punkt przecięcia z osią $ y $),
- $ \beta_1 $ jest współczynnikiem kierunkowym (nachylenie prostej),
- $ \epsilon $ to błąd (reszta), czyli różnica pomiędzy wartością rzeczywistą a przewidywaną.

### **Metoda najmniejszych kwadratów**

Metoda najmniejszych kwadratów polega na znalezieniu takich wartości $ \beta_0 $ i $ \beta_1 $, które minimalizują sumę kwadratów odchyleń wartości przewidywanych od rzeczywistych. Matematycznie jest to wyrażone jako:

$$
\min_{\beta_0, \beta_1} \sum_{i=1}^{n} \left( y_i - (\beta_0 + \beta_1 x_i) \right)^2
$$

Aby znaleźć optymalne wartości $ \beta_0 $ i $ \beta_1 $, możemy skorzystać z następujących wzorów:

$$
\beta_1 = \frac{\sum_{i=1}^{n} (x_i - \overline{x})(y_i - \overline{y})}{\sum_{i=1}^{n} (x_i - \overline{x})^2}
$$

$$
\beta_0 = \overline{y} - \beta_1 \overline{x}
$$

gdzie:

- $ \overline{x} $ to średnia arytmetyczna zmiennych $ x $,
- $ \overline{y} $ to średnia arytmetyczna zmiennych $ y $.

**Treść zadania:**

Wprowadź współczynnik determinacji $R^2$, który mierzy, jak dobrze model regresji wyjaśnia zmienność danych.

**Definicja:**

Współczynnik determinacji $R^2$ jest miarą jakości dopasowania modelu regresji liniowej. Informuje, jaka część całkowitej zmienności zmiennej zależnej $y$ jest wyjaśniona przez zmienną niezależną $x$ w modelu.

Wartość $R^2$ zawiera się w przedziale od 0 do 1:
- $R^2 = 1$ oznacza idealne dopasowanie modelu do danych (wszystkie punkty leżą na prostej regresji).
- $R^2 = 0$ oznacza, że model nie wyjaśnia zmienności danych lepiej niż średnia $\bar{y}$.
- Wartości ujemne mogą wystąpić, gdy model jest źle dopasowany (np. gdy nie zawiera wyrazu wolnego).

**Wzór na współczynnik determinacji:**

$$
R^2 = 1 - \frac{\mathrm{SS}_{\mathrm{res}}}{\mathrm{SS}_{\mathrm{tot}}}
$$

gdzie:
- $\mathrm{SS}_{\mathrm{res}} = \sum_{i=1}^n (y_i - \hat{y}_i)^2$ - suma kwadratów reszt (błąd modelu),
- $\mathrm{SS}_{\mathrm{tot}} = \sum_{i=1}^n (y_i - \bar{y})^2$ - całkowita suma kwadratów (całkowita zmienność danych),
- $y_i$ - rzeczywiste wartości zmiennej zależnej,
- $\hat{y}_i$ - wartości przewidywane przez model regresji,
- $\bar{y}$ - średnia wartość zmiennej zależnej $y$.
